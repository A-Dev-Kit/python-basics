## Q28  Character Compression

**Hint:** Build a compressed string with counts of consecutive characters.

```python
text = "aaabbcdddd"
# Compress to format like a3b2c1d4

# Your solution below:
result = ''
count = 1
for index in range(1, len(text)):
    if text[index] == text[index - 1]:
        count += 1
    else:
        result += text[index - 1]
        result += str(count)
        count = 1
result += text[len(text) - 1]
result += str(count)
print(result)
```

## Q45  Merging Nested Dicts: Shallow vs Deep

**Hint:** `dict.update` performs a shallow merge — show consequences with nested dicts and how to perform a deep merge.

```python
a = {'x': {'n': 1}}
b = {'x': {'m': 2}}
res = a.copy()
res.update(b)
# Explain what `res['x']` contains and why; suggest a deep merge approach

# Your solution below:
# update is shallow - b's value for 'x' replaces a's
print(res['x'])
print('n' in res['x'])

deep_res = a.copy()
deep_res['x'] = a['x'].copy()
deep_res['x'].update(b['x'])
deep_res['x']
```
