##### Sets

```python
set_techs = {1,2,3,4,5,1}
set_techs
```

```python
set_techs_new = {5,6,7,8,9}
```

```python
set_combined = set_techs.union(set_techs_new)
set_combined
```

```python
set_combined_new = set_techs_new.union(set_techs) # sets maintaining sorted order
set_combined_new
```

```python
set_techs.add(12)
print(set_techs)
```

```python
set4 = set_combined_new.difference(set_techs)
set4
```

```python
set5 = set_techs.intersection(set_techs_new)
set5
```
